## Problem 5 - Numerical Solution of ODEs
In this problem we will implement Euler’s method, second order Taylor’s method, and Heun’s method, and use them to approximate the solution to the ODE,

$$
y' = -2ty, \qquad y(0) = 1
$$

The exact solution to this equation is $y(t) = e^{-t^{2}}$.

```python
def y_exact(t):
    """
    Computes the exact solution of the ODE y'(t) = -2ty with y(t) = exp(-t^2).

    Parameters
    ----------
    t : float or array-like
        The time values at which to compute the exact solution.

    Returns
    -------
    float or array-like
        The exact solution y(t) = exp(-t^2).
    """
    return np.exp(-(t**2))


def f(t, y):
    """
    Computes the derivative f(t, y) = -2 * t * y for the given ODE y'(t) = -2ty.

    Parameters
    ----------
    t : float
        The current time value.
    y : float
        The current value of y.

    Returns
    -------
    float
        The value of the derivative at (t, y).
    """
    return -2 * t * y

def f_t(t, y):
    """
    Computes the total derivative of f(t, y) = -2ty with respect to t for the second-order Taylor method.

    Parameters
    ----------
    t : float
        The current time value.
    y : float
        The current value of y.

    Returns
    -------
    float
        The total derivative of f with respect to t.
    """
    return -2*y + 4*(t**2)*y

def _num_steps(t_0, t_end, h):
    """
    Computes the number of steps needed for a given step size h over the time interval [t_0, t_end].

    Parameters
    ----------
    t_0 : float
        The initial time.
    t_end : float
        The final time.
    h : float
        The step size.

    Returns
    -------
    int
        The number of steps needed.
    """
    return int((t_end - t_0) / h) + 1


def euler(h, f, y_0, t_0, t_end):
    """
    Solves an ODE using Euler's method.

    Parameters
    ----------
    h : float
        The step size.
    f : function
        The derivative function f(t, y).
    y_0 : float
        The initial value of y at t_0.
    t_0 : float
        The initial time.
    t_end : float
        The final time.

    Returns
    -------
    t_values : numpy.ndarray
        Array of time points at which the solution is computed.
    y_values : numpy.ndarray
        Array of the approximate solution values at each time point.
    """
    # Calculate number of steps
    num_steps = _num_steps(t_0=t_0, t_end=t_end, h=h)
    
    # Initialize t and y values
    t_values = np.linspace(start=t_0, stop=t_end, num=num_steps)
    y_values = np.zeros(t_values.shape)

    # Set y0
    y_values[0] = y_0

    for i in range(1, num_steps):
        y_values[i] = y_values[i-1] + h*f(t_values[i-1], y_values[i-1])

    return t_values, y_values


def second_order_taylor(h, f, f_t, y_0, t_0, t_end):
    """
    Solves an ODE using the second-order Taylor method.

    Parameters
    ----------
    h : float
        The step size.
    f : function
        The derivative function f(t, y).
    f_t : function
        The total derivative of f with respect to t.
    y_0 : float
        The initial value of y at t_0.
    t_0 : float
        The initial time.
    t_end : float
        The final time.

    Returns
    -------
    t_values : numpy.ndarray
        Array of time points at which the solution is computed.
    y_values : numpy.ndarray
        Array of the approximate solution values at each time point.
    """
    # Calculate number of steps
    num_steps = _num_steps(t_0=t_0, t_end=t_end, h=h)

    # Initialize t and y values
    t_values = np.linspace(start=t_0, stop=t_end, num=num_steps)
    y_values = np.zeros(t_values.shape)

    # Set y0
    y_values[0] = y_0

    for i in range(1, num_steps):
        y_n = y_values[i-1]
        t_n = t_values[i-1]

        y_values[i] = y_n + h*f(t_n, y_n) + ((h**2)/2) * f_t(t_n, y_n)

    return t_values, y_values


def heuns(h, f, y_0, t_0, t_end):
    """
    Solves an ODE using Heun's method (a second-order Runge-Kutta method).

    Parameters
    ----------
    h : float
        The step size.
    f : function
        The derivative function f(t, y).
    y_0 : float
        The initial value of y at t_0.
    t_0 : float
        The initial time.
    t_end : float
        The final time.

    Returns
    -------
    t_values : numpy.ndarray
        Array of time points at which the solution is computed.
    y_values : numpy.ndarray
        Array of the approximate solution values at each time point.
    """
    # Calculate number of steps
    num_steps = _num_steps(t_0=t_0, t_end=t_end, h=h)

    # Initialize t and y values
    t_values = np.linspace(start=t_0, stop=t_end, num=num_steps)
    y_values = np.zeros(t_values.shape)

    # Set y0
    y_values[0] = y_0

    for i in range(1, num_steps):
        y_n = y_values[i-1]
        t_n = t_values[i-1]

        f_n = f(t_n, y_n)

        y_star = y_n + h*f_n

        y_values[i] = y_n + (h/2) * (f_n + f(t_n+h, y_star))

    return t_values, y_values
```
